```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("raw-captures/face-dim.jpg")
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
print(image.shape)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
```

```python
H, S, V = cv2.split(hsv)
print(H)
print(S)
print(V)
```

```python
plt.figure(figsize=(15,10))

plt.subplot(231)
plt.imshow(image_rgb)
plt.title("RGB")
plt.axis("off")

plt.subplot(232)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(233)
plt.imshow(H, cmap="gray")
plt.title("Hue Channel")
plt.axis("off")

plt.subplot(234)
plt.imshow(S, cmap="gray")
plt.title("Saturation Channel")
plt.axis("off")

plt.subplot(235)
plt.imshow(V, cmap="gray")
plt.title("Value Channel")
plt.axis("off")

plt.tight_layout()
plt.show()
```

After conversion of the images, I observed that

Grayscale Image

    The grayscale image retained most of the visible details from the original low-light image. Although all color information was removed, the intensity differences were sufficient to preserve the shapes and textures of objects.
    
Hue (H) Channel

    The Hue channel appeared noisy, especially in the darker regions. In low-light conditions, color information becomes less reliable because the camera sensor receives less light, making hue values unstable and less useful for preserving details.

Saturation (S) Channel

    The Saturation channel was slightly better than the Hue channel but still contained noticeable noise. Since the image was captured in low light, many regions had weak or desaturated colors, reducing the amount of useful information in this channel.
    
Value (V) Channel

    The Value channel preserved the most detail among the HSV channels. Since the Value channel represents the brightness (intensity) of each pixel, it retained the variations in light and shadow, making object boundaries and textures more visible even under low-light conditions.
