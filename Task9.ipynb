```python
import cv2
```

```python
cap = cv2.VideoCapture("raw-captures/object-video.mp4")
print(cap.isOpened())
```

```python
fps = cap.get(cv2.CAP_PROP_FPS)

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))

height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

print(f"FPS: {fps}")
print(f"Width: {width}")
print(f"Height: {height}")
```

```python
fourcc = cv2.VideoWriter_fourcc(*'mp4v')

out = cv2.VideoWriter(
    "edge_output.mp4",
    fourcc,
    fps,
    (width, height),
    isColor=False
)
```

```python
while True:

    ret, frame = cap.read()

    if not ret:
        break
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    gray = cv2.equalizeHist(gray)

    blur = cv2.GaussianBlur(gray, (5,5), 0)

    edges = cv2.Canny(blur, 50, 150)
    out.write(edges)

cap.release()
out.release()
```

Before applying histogram equalization and Gaussian Blur, the edge detector missed several parts of the moving object, resulting in broken and faint edges. After using cv2.equalizeHist() and GaussianBlur(), the contrast between the object and the background increased, making the object boundaries more prominent. Consequently, Canny edge detection produced cleaner, sharper, and more complete outlines throughout the video.
